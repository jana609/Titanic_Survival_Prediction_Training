data handling

```python
import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

train_data = pd.read_csv('C:\\Users\\jna\\Desktop\\titanic project\\titanic\\train.csv')

#original data statistics
print('head : ', train_data.head())
print('info : ',train_data.info())
print('describe : ',train_data.describe)

# remove unuseful col
train_data.drop(['PassengerId', 'Name', 'Ticket', 'Cabin', 'Embarked'], axis=1, inplace=True)
print('info after dropping : ',train_data.info())

#handling missing values using median for age
train_data['Age'] = train_data['Age'].fillna(train_data.groupby('Pclass')['Age'].transform('median'))
print('info after handling missing values : ',train_data.info())
print('new ages : ',train_data.Age)

#encode male and female
train_data['Sex'] = train_data['Sex'].map({'male': 0, 'female': 1})

# x and y
y = train_data['Survived']
x = train_data.drop('Survived', axis=1)

print("X:", x.shape)
print("y:", y.shape)

#splitting data
x_training ,x_testing , y_training,y_testing  = train_test_split(x ,y ,test_size=0.2 ,stratify=y ,random_state=42)
print("X_train shape:", x_training.shape)
print("X_test shape:", x_testing.shape)

#scaling
scalar = StandardScaler()
scaled_x_training = scalar.fit_transform(x_training)
scaled_x_testing = scalar.transform(x_testing)
print("scaled X_train shape:", scaled_x_training)
print("scaled X_test shape:", scaled_x_testing)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(scaled_x_training ,y_training )
y_pred_knn= knn.predict(scaled_x_testing)

accuracy = accuracy_score(y_testing , y_pred_knn)
print (accuracy)

###################################################

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(scaled_x_training ,y_training )
y_pred_knn= knn.predict(scaled_x_testing)

accuracy = accuracy_score(y_testing , y_pred_knn)
print (accuracy)

###################################################3

knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(scaled_x_training ,y_training )
y_pred_knn= knn.predict(scaled_x_testing)

accuracy = accuracy_score(y_testing , y_pred_knn)
print (accuracy)

######################################################

knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(scaled_x_training ,y_training )
y_pred_knn= knn.predict(scaled_x_testing)

accuracy = accuracy_score(y_testing , y_pred_knn)
print (accuracy)
```

3 is most accurate

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(scaled_x_training ,y_training )
y_predict= knn.predict(scaled_x_testing)

accuracy_knn = accuracy_score(y_testing , y_predict)
print ('accuracy for knn : ',accuracy_knn)

precision_knn = precision_score(y_testing , y_predict)
print ('precision for knn : ',precision_knn)

recall_knn = recall_score(y_testing , y_predict)
print ('recall for knn : ',recall_knn)

f1_knn = f1_score(y_testing , y_predict)
print ('f1 score for knn : ' ,f1_knn)

cm_knn = confusion_matrix(y_testing , y_predict)
print ('confusion_matrix for knn : \n' ,cm_knn)
```

decision tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_training ,y_training )
y_pred_dt= dt.predict(x_testing)

accuracy_dt = accuracy_score(y_testing , y_pred_dt)
print ('accuracy for decision tree : ',accuracy_dt)

precision_dt = precision_score(y_testing , y_pred_dt)
print ('precision for decision tree : ',precision_dt)

recall_dt = recall_score(y_testing , y_pred_dt)
print('recall for decision tree : ' ,recall_dt)

f1_dt = f1_score(y_testing , y_pred_dt)
print ('f1 score for decision tree : ',f1_dt)

cm_dt = confusion_matrix(y_testing , y_pred_dt)
print ('confusion_matrix for decision tree : \n',cm_dt)
```

Compare the unrestricted tree with at least two max_depth values and observe the effect on train and test
accuracy.

comparison

```python
y_pred_dt3 = dt_depth_3.predict(x_testing)
def compare_accuracy(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf):
    scores = {
        "KNN": accuracy_score(y_testing, y_pred_knn),
        "Decision Tree": accuracy_score(y_testing,y_pred_dt3),
        "Random Forest": accuracy_score(y_testing, y_pred_rf)
    }
    best_model = max(scores, key=scores.get)
    print("Best Accuracy : " ,best_model)


def compare_precision(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf):
    scores = {
        "KNN": precision_score(y_testing, y_pred_knn),
        "Decision Tree": precision_score(y_testing,y_pred_dt3),
        "Random Forest": precision_score(y_testing, y_pred_rf)
    }
    best_model = max(scores, key=scores.get)
    print("Best Precision : ",best_model)


def compare_recall(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf):
    scores = {
        "KNN": recall_score(y_testing, y_pred_knn),
        "Decision Tree": recall_score(y_testing,y_pred_dt3),
        "Random Forest": recall_score(y_testing, y_pred_rf)
    }
    best_model = max(scores, key=scores.get)
    print("Best Recall : " ,best_model)


def compare_f1(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf):
    scores = {
        "KNN": f1_score(y_testing, y_pred_knn),
        "Decision Tree": f1_score(y_testing,y_pred_dt3),
        "Random Forest": f1_score(y_testing, y_pred_rf)
    }
    best_model = max(scores, key=scores.get)
    print("Best F1-Score : ",best_model)

compare_accuracy(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf)
compare_precision(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf)
compare_recall(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf)
compare_f1(y_testing, y_pred_knn,y_pred_dt3, y_pred_rf)


```
